```python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="microsoft/Phi-3-mini-4k-instruct",
    device_map="auto"
)
```

```python
while True:
    user_input = input("You: ")
    if user_input == "exit":
        break
    result = generator(user_input, max_new_tokens=100)
    print("Bot:", result[0]["generated_text"])
```

```python
conversation = []

while True:
    user_input = input("You: ")
    if user_input == "exit":
        break
    conversation.append({"role": "user", "content": user_input})

    result = generator(conversation, max_new_tokens=100)
    bot_reply = result[0]["generated_text"][-1]["content"]

    conversation.append({"role": "assistant", "content": bot_reply})
    print("Bot:", bot_reply)
```
