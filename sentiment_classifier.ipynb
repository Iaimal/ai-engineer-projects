```python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="microsoft/Phi-3-mini-4k-instruct",
    device_map="auto"
)

text_to_summarize = """
Paste any paragraph or article here that you want summarized.
For example: a news article, a Wikipedia paragraph, or your own notes.
"""

prompt = f"Summarize this in 3 bullet points:\n\n{text_to_summarize}"
result = generator(prompt, max_new_tokens=150)
print(result[0]["generated_text"])
```

```python
review = "The package arrived on Tuesday as scheduled."

prompt = f"Classify the sentiment of this text as exactly one word: Positive, Negative, or Neutral.\n\nText: {review}\n\nSentiment:"

result = generator(prompt, max_new_tokens=10)
print(result[0]["generated_text"])
```
