```python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="microsoft/Phi-3-mini-4k-instruct",
    device_map="auto"
)
```

```python
while True:
    print("\nChoose an option:")
    print("1. Summarize text")
    print("2. Classify sentiment")
    print("3. Chat")
    print("4. Exit")

    choice = input("Enter choice (1-4): ")

    if choice == "1":
        text = input("Enter text to summarize: ")
        prompt = f"Summarize this text in 3 bullet points:\n\n{text}\n\nSummary:"
        result = generator(prompt, max_new_tokens=100)
        print(result[0]["generated_text"])

    elif choice == "2":
        review = input("Enter text to classify: ")
        prompt = f"Classify the sentiment of this text as exactly one word: Positive, Negative, or Neutral.\n\nText: {review}\n\nSentiment:"
        result = generator(prompt, max_new_tokens=10)
        print(result[0]["generated_text"])

    elif choice == "3":
        print("Entering chat mode. Type 'exit' to return to the main menu.")
        conversation = []
        while True:
            user_input = input("You: ")
            if user_input == "exit":
                break
            conversation.append({"role": "user", "content": user_input})
            result = generator(conversation, max_new_tokens=100)
            bot_reply = result[0]["generated_text"][-1]["content"]
            conversation.append({"role": "assistant", "content": bot_reply})
            print("Bot:", bot_reply)

    elif choice == "4":
        print("Goodbye!")
        break

    else:
        print("Invalid choice, try again.")
```
