# Character-Level Tokenizer (nanoGPT foundations)

Built while studying transformer/LLM internals (following the ideas behind Andrej Karpathy's nanoGPT).

Implements:
- Character-level `encode` / `decode` functions
- Conversion to a PyTorch tensor
- Train / validation split
- Input/target windowing for next-token prediction (the core training signal behind language models)

```python
# Tokenization
text = "hello world"
unique_chars = set(text)
print(unique_chars)

char_to_num = {}
for index, char in enumerate(unique_chars):
    char_to_num[char] = index
print(char_to_num)

num_to_char = {}
for index, char in enumerate(unique_chars):
    num_to_char[index] = char
print(num_to_char)
```

```python
# Encoder: text -> list of numbers
def encode(s):
    result = []
    for char in s:
        result.append(char_to_num[char])
    return result

print(encode("world"))
```

```python
# Decoder: list of numbers -> text
def decode(numbers):
    result = []
    for num in numbers:
        result.append(num_to_char[num])
    return "".join(result)

print(decode([4, 7, 0, 0, 3]))
```

```python
# Convert the whole text into a PyTorch tensor
import torch

data = torch.tensor(encode(text))
print(data)
```

```python
print(len(text))
```

```python
# Train / validation split (90% / 10%)
n = int(0.9 * len(data))
print(n)
```

```python
train_data = data[:n]
val_data = data[n:]
print(train_data)
print(val_data)
```

```python
# Input/target windowing: y is x shifted forward by one position.
# This is the core training signal for next-token prediction --
# at every position, the target is "what character comes next".
block_size = 4
x = train_data[:block_size]
y = train_data[1:block_size + 1]
print(x)
print(y)
```
